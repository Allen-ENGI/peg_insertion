```python
import gymnasium as gym
import metaworld
import numpy as np
import matplotlib.pyplot as plt
from IPython import display
import time

camera_names = ['corner', 'corner2', 'corner3', 'topview', 'behindGripper', 'gripperPOV']

# 1. Initialize environment with offscreen rendering
env = gym.make(
    "Meta-World/MT1", 
    env_name="peg-insert-side-v2", 
    render_mode="rgb_array", 
    camera_name="corner2"
)

obs, info = env.reset()

# 2. Setup figure for dynamic in-notebook playback (flip the initial frame)
fig, ax = plt.subplots(figsize=(6, 6))
initial_frame = np.flipud(env.render())
img_display = ax.imshow(initial_frame)
plt.axis("off")
plt.title("MuJoCo Physics: Upright Sawyer Arm (peg-insert-side-v3)")

# 3. Step through MuJoCo physics with continuous motion commands
for step in range(80):
    action = np.array([
        0.5 * np.sin(step * 0.15),   # X movement
        0.5 * np.cos(step * 0.15),   # Y movement
        0.2 * np.sin(step * 0.3),    # Z movement
        0.0                          # Gripper state
    ], dtype=np.float32)
    
    obs, reward, terminated, truncated, info = env.step(action)
    
    # Grab frame, flip vertically, and update display
    frame = np.flipud(env.render())
    img_display.set_data(frame)
    display.display(plt.gcf())
    display.clear_output(wait=True)
    
    time.sleep(0.01)

env.close()
print("MuJoCo physics loop completed successfully with upright rendering!")
```
